# Restaurant Dataset : Data Cleaning

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../Dataset.csv")

df.head()
```

```python
clean_df = df.copy()
```

```python
missing = clean_df.isnull().sum()

missing[missing > 0].sort_values(ascending=False)
```

```python
clean_df["Cuisines"] = clean_df["Cuisines"].fillna(clean_df["Cuisines"].mode()[0])
```

```python
clean_df.isnull().sum()
```

```python
clean_df.isnull().sum()
```

```python
clean_df.dtypes
```

```python
clean_df["Restaurant ID"] = clean_df["Restaurant ID"].astype(str)
```

```python
clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop_duplicates()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df.describe().T
```

```python
print(clean_df.shape)

print(clean_df.isnull().sum().sum())

print(clean_df.duplicated().sum())

clean_df.info()
```

```python
clean_df.to_csv("Cleaned_Dataset.csv", index=False)
```

## Cleaning Summary

- Missing values handled.
- Data types verified.
- Duplicate rows removed.
- Dataset validated.
- Cleaned dataset saved as `Cleaned_Dataset.csv`.

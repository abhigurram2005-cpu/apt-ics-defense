# Restaurant Dataset - Feature Engineering

### Level 2 - Task 3

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
## Load the cleaned dataset
df = pd.read_csv("../Cleaned_Dataset.csv")

df.head()
```

### Text Based Features

```python
#restaurant name length
df["Restaurant Name Length"] = df["Restaurant Name"].str.len()

df[["Restaurant Name", "Restaurant Name Length"]].head()
```

```python
#address length
df["Address Length"] = df["Address"].str.len()

df[["Address", "Address Length"]].head()
```

```python
#Number of cuisines offered by each restaurant
df["Cuisine Count"] = (
    df["Cuisines"]
    .fillna("")
    .str.split(",")
    .apply(len)
)

df[["Cuisines", "Cuisine Count"]].head()
```

### Binary Encodinng

```python
#Encode table booking feature
df["Has Table Booking"] = (
    df["Has Table booking"]
    .map({"Yes": 1, "No": 0})
)

df[["Has Table booking", "Has Table Booking"]].head()
```

```python
#Encode online delivery feature
df["Has Online Delivery"] = (
    df["Has Online delivery"]
    .map({"Yes": 1, "No": 0})
)

df[["Has Online delivery", "Has Online Delivery"]].head()
```

```python
# Categorize restaurants based on their aggregate rating
bins = [-1, 2.5, 3.5, 4.5, 5]
labels = ["Poor", "Average", "Good", "Excellent"]

df["Rating Category"] = pd.cut(
    df["Aggregate rating"],
    bins=bins,
    labels=labels
)

df[["Aggregate rating", "Rating Category"]].head()
```

### Optional Engineered features

```python
df["Cost Per Vote"] = (
    df["Average Cost for two"] /
    (df["Votes"] + 1)
)

df[["Average Cost for two", "Votes", "Cost Per Vote"]].head()
```

```python
df["Has Votes"] = (df["Votes"] > 0).astype(int)

df[["Votes", "Has Votes"]].head()
```

```python

new_features = [
    "Restaurant Name Length",
    "Address Length",
    "Cuisine Count",
    "Has Table Booking",
    "Has Online Delivery",
    "Cost Per Vote",
    "Has Votes",
    "Rating Category"
]

df[new_features].head()
```

```python
#save this engineered dataset to a new CSV file
df.to_csv("Engineered_Dataset.csv", index=False)
```

## Feature Engineering Summary

- Extracted text-based features from restaurant names and addresses.
- Counted the number of cuisines served by each restaurant.
- Encoded table booking and online delivery into binary features.
- Created additional features to support predictive modeling.
- Saved the processed dataset as `Engineered_Dataset.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../Cleaned_Dataset.csv")

df.head()
```

```python
df.describe().T
```

```python
df.mean(numeric_only=True)
```

```python
df.median(numeric_only=True)
```

```python
# Descriptive Statistics Table

stats = (
    df[["Aggregate rating", "Votes", "Average Cost for two", "Price range"]]
    .describe()
    .round(2)
)

stats
```

```python
df.mode().iloc[0]
```

```python
df.std(numeric_only=True)
```

The numerical features show the central tendency and spread of restaurant ratings, votes, prices, and other quantitative attributes.

```python
country_counts = df["Country Code"].value_counts()

country_counts
```

```python
plt.figure(figsize=(8,5))

country_counts.head(10).plot(kind="bar")

plt.title("Top Countries by Number of Restaurants")
plt.xlabel("Country Code")
plt.ylabel("Restaurant Count")

plt.show()
```

```python
city_counts = df["City"].value_counts()

city_counts
```

```python
city_counts.head(10)
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x=city_counts.head(10).values,
    y=city_counts.head(10).index
)

plt.title("Top 10 Cities")
plt.xlabel("Restaurants")

plt.show()
```

```python
cuisine_counts = (
    df["Cuisines"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
)

cuisine_counts
```

```python
cuisine_counts.head(10)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=cuisine_counts.head(10).values,
    y=cuisine_counts.head(10).index
)

plt.title("Top 10 Cuisines")
plt.xlabel("Restaurant Count")

plt.show()
```

```python
chain_counts = df["Restaurant Name"].value_counts()

chain_counts[chain_counts > 1].head(20)
```

```python
df[
    ["Restaurant Name", "Votes"]
].sort_values(
    by="Votes",
    ascending=False
).head(10)
```

```python
print("Most Common Country:")
print(country_counts.idxmax())

print()

print("Most Common City:")
print(city_counts.idxmax())

print()

print("Most Common Cuisine:")
print(cuisine_counts.idxmax())
```

```python
print("Most Common Country:")
print(country_counts.idxmax())

print()

print("Most Common City:")
print(city_counts.idxmax())

print()

print("Most Common Cuisine:")
print(cuisine_counts.idxmax())
```

## Key Findings

- The dataset contains restaurants from multiple countries.
- A few cities contribute the majority of restaurants.
- Some cuisines dominate the dataset.
- Several restaurant chains have multiple branches.
- Customer votes are concentrated among a small number of restaurants.
